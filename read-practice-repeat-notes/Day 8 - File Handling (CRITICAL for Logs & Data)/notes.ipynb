```python
# Reading a File
with open("file.txt", "r") as f:
    data = f.read()

print(data)
```

```python
# Read line by line (IMPORTANT)
with open("file.txt", "r") as f:
    for line in f:
        print(line.strip())
```

```python
# Writing to file
with open("file.txt", "w") as f:
    f.write("Hello World\n")
```

```python
# Append (don’t overwrite)
with open("output.txt", "a") as f:
    f.write("New line\n")
```

```python
# DevOps Example (Log Parsing)
with open("access.log", "r") as f:
    for line in f:
        if "500" in line:
            print(line)
```
