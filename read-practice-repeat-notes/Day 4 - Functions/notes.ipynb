##### Concept: Functions
##### What it does: Reusable blocks of code

##### Basic function
```
def greet():
    print("Hello")
greet()
```
##### With arguments
```
def greet(name):
    print(f"Hello {name}")
greet("DevOps")
```
##### Return value
```
def add(a, b):
    return a + b
result = add(2, 3)
print(result)
```
##### Default argument
```
def deploy(service, env="dev"):
    print(f"{service} -> {env}")
deploy("api")
deploy("api", "prod")
```

📓 NOTEBOOK NOTE — Function Inputs vs Globals
-----------------------------------
##### Concept: Function Inputs
##### What it does: Makes function reusable and predictable

##### BAD (depends on global variable)
memory = 90
def check():
    return memory

##### GOOD (takes input)
def check(memory):
    return memory

##### Key takeaway: Always pass values into functions
##### Practice: Refactor old code to remove global dependencies
##### Time suggestion: 20 min

````python
def check_cpu(cpu):
    if cpu > 80:
        return "High"
    elif cpu > 50:
        return "Moderate"
    else:
        return "Low"

result = check_cpu(75)
print(result)
````

````python
# Multiple Returns (useful)
def get_metrics():
    cpu = 70
    memory = 60
    return cpu, memory

c, m = get_metrics()
print(c, m)
````

````python
# Default Arguments
def greet(name="User"):
    print(f"Hello {name}")

greet()        # Hello User
greet("Krish") # Hello Krish
````

````python
# Return Values (CRITICAL)
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
````

````python
def get_status(cpu):
    if cpu > 70:
        return "Scale"
    else:
        return "Not Required"
get_status(71)
````
