### Concept: Loops (for, while)
### What it does: Repeats tasks
#### for loop
```for i in range(3):
    print(i)
```
#### range with start/end
```
for i in range(1, 4):
    print(i)
```
#### loop through string
```
for c in "ai":
    print(c)
```
#### while loop
```
count = 0
while count < 3:
    print(count)
    count += 1
```

#### break
```
for i in range(5):
    if i == 3:
        break
    print(i)
```
#### continue
```
for i in range(5):
    if i == 2:
        continue
    print(i)
```

````python
# for loop (Most Important)
# Used when you iterate over a collection

for i in range(5):
    print(i)
````

````python
# while loop
count = 0

while count < 3:
    print(count)
    count += 1

# Careful: infinite loops if you forget increment
````

````python
# break, continue and pass
for i in range(5):
    if i == 3:
        break
    print(i)

for i in range(5):
    if i == 2:
        continue
    print(i)

for i in range(5):
    if i == 3:
        pass
    print(i)
````

````python
servers = ["web1", "web2", "db1"]

for server in servers:
    print(f"Pinging {server}")

count = 1
while count <= 3:
    print(f"Retry {count}")
    count += 1
````

##### for → fixed number of times run (how many times)
##### while → run until condition becomes false (when to stop)
##### break → stop the loop immediately when condition met
##### continue → skip current iteration when condition met
##### pass → do nothing, just placeholder (code runs normally)
