##### Concept: Lists
##### What it does: Stores multiple values in one variable
```
servers = ["web1", "web2"]
```
##### access
```
print(servers[0])
```
##### add
```
servers.append("db1")
```
##### remove
```
servers.remove("web2")
```
##### loop
```
for s in servers:
    print(s)
```
##### Key takeaway: Lists store collections (servers, logs, metrics)

````python
# Concept: List
# A list is a collection of items
servers = ["web1", "web2", "db1"]

#Like: servers=(web1 web2 db1)
````

````python
# Access elements (indexing)
print(servers[0])  # web1
print(servers[1])  # web2

# Index starts from 0
````

````python
# Loop through list
for server in servers:
    print(server)
````

````python
# Add elements
servers.append("cache1")

# Remove elements
servers.remove("db1")

# Length
print(len(servers))

# Mixed types (allowed but avoid)
data = ["web1", 80, True]
````

````python
servers = ["web1", "web2", "db1"]

# add new server
servers.append("cache1")

# remove db1
servers.remove("db1")

# print all
for s in servers:
    print(f"Server: {s}")

print(f"Total servers: {len(servers)}")
````

##### Concept: Lists, Dicts, Sets, Tuples
##### What it does: Store collections of data

##### List
```
services = ["api", "auth"]
services.append("db")
```
##### Dict
```
config = {"env": "prod", "replicas": 2}
```
##### Tuple
```
coords = (10, 20)
```
##### Set
```
ids = {1, 2, 2, 3}
```
##### Loop list
```
for s in services:
    print(s)
```
##### Loop dict
```
for k, v in config.items():
    print(k, v)
```
