````python
# Concept: if / else
cpu = 85

if cpu > 80:
    print("High CPU usage")
else:
    print("CPU is normal")
````

````python
# Multiple Conditions (elif)
cpu = 65

if cpu > 80:
    print("High")
elif cpu > 50:
    print("Moderate")
else:
    print("Low")
````

#### Logical Operators

| Operator | Meaning                      |
| -------- | ---------------------------- |
| `and`    | both conditions must be true |
| `or`     | at least one true            |
| `not`    | reverse condition            |

```
cpu = 85
memory = 70

if cpu > 80 and memory > 60:
    print("High load on system")
```

````python
# Multiple conditions
cpu = 82
memory = 45

if cpu > 80:
    print("CPU is high")
elif cpu > 50:
    print("CPU is moderate")
else:
    print("CPU is low")

if cpu > 80 and memory < 50:
    print("CPU high but memory OK")
````
