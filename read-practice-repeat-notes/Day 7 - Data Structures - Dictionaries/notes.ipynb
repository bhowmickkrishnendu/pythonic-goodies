````python
# Concept: Dictionary
# Key → Value mapping
server = {
    "name": "web1",
    "cpu": 75,
    "status": "running"
}

# Access values
print(server["name"])   # web1
print(server["cpu"])    # 75

# Add / update values
server["memory"] = 60
server["cpu"] = 80

# Loop through dictionary
for key, value in server.items():
    print(f"{key}: {value}")

# Real DevOps Example
servers = [
    {"name": "web1", "cpu": 70},
    {"name": "web2", "cpu": 85}
]

for s in servers:
    if s["cpu"] > 80:
        print(f"{s['name']} is overloaded")
````

````python
server = {
    "name": "web1",
    "cpu": 75
}

# add memory
server["memory"] = 60

# update cpu
server["cpu"] = 85

# print all
for k, v in server.items():
    print(f"{k}: {v}")
````

##### Concept: Dictionary (key-value)
##### What it does: Stores structured data like JSON
```
server = {
    "name": "web1",
    "cpu": 75
}
```
##### access
print(server["name"])

##### update
```
server["cpu"] = 80
```
##### loop
```
for k, v in server.items():
    print(k, v)
```
##### When to use loop (important distinction)

```
servers = [
    {"name": "web1", "cpu": 70},
    {"name": "web2", "cpu": 90}
]

for s in servers:
    if s["cpu"] > 80:
        print(f"{s['name']} is overloaded")
```

##### Key Insight
```
dict → access directly → server["cpu"]
list of dicts → loop → for s in servers
```

##### Dict vs List of Dicts
##### Single dictionary
```
server = {"name": "web1", "cpu": 90}
print(server["cpu"])
```
##### List of dictionaries
```
servers = [
    {"name": "web1", "cpu": 70},
    {"name": "web2", "cpu": 90}
]

for s in servers:
    print(s["name"])
```
##### Key takeaway:
##### dict = one object
##### list of dicts = multiple objects
